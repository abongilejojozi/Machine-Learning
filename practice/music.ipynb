```python
import pandas as pd 

from sklearn.tree import DecisionTreeClassifier 
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score 

# importing the data: 
music_data = pd.read_csv("music.csv") 


#Cleaning the data, and also preparing the data as input and output. 

X = music_data.drop(columns=["genre"]) #input set  
y = music_data["genre"] #output set 

#and then I need to call the train_test_split function to destructure my data according to a certain test size. 

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2) #this function gives me the input to output split for training and testing my model. 
#the train_test_split function split the data in the correct precision but randomly. 
#Then I create my model. 
model = DecisionTreeClassifier() # Creating the DecisionTreeClassifier object as our model here. 
#and then we can start training the model. 
model.fit(X_train, y_train) #Allow the model to train and get fit to make predictions. 

#test it ... allow it to predict output and compare to the real output. 
predictions = model.predict(X_test) 
predictions #I give the model the input set I use for testing and let it predict. 


#and then I have to check how accurate is it. Doing some error analysis, or scoring to check how far it is from the actual truth. 
score = accuracy_score(y_test, predictions)
score #I need to remember to always inspect the output here. 
```
