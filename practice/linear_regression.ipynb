```python
#my data 
import pandas as pd 
import numpy as np 

#model imports 
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import r2_score, mean_squared_error 

#for visualising the data 
import matplotlib.pyplot as plt 
import seaborn as sns 


data = pd.read_csv("BostonHousing.csv")
data.head()
```

```python
X = data.drop(columns=["medv"]) #This is my input set  
y = data["medv"] #This is my output set 

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2) 
print("X data: ", X.shape)
print("X_train: ", X_train.shape)
```

```python
print("y data: ", y.shape) 
print("y_train: ", y_train.shape) 
```

```python
#Now we create a model and then train it on this data . 

model = LinearRegression() 
model.fit(X_train, y_train) 
```

```python
predictions = model.predict(X_test) 
predictions 
```

```python
#Check accuracy. 
MSE = mean_squared_error(y_test, predictions)

print("Mean Square Error: %.2f" % MSE)
```

```python
R2 = r2_score(y_test, predictions)
print("R2 score: %.2f" % R2)
```

```python
#finally visualise the data. 

sns.scatterplot(x = y_test, y = predictions, marker="+", alpha=0.5)

plt.xlabel("Actual") 
plt.ylabel("Predicted") 
plt.title("Actual vs Predicted")
```
